# Algorithmic Native Speakerism: V-Shaped Trajectory Replication
This notebook provides the computational analysis for the V-shaped pattern in lexical diversity (MTLD). It loads the replication dataset, computes stage-wise means, and generates the empirical plots.


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset from the relative path as defined in the repo structure
# Stage labels: Original (OD), AI_Revised (AIR), Author_Refined (AR)
try:
    df = pd.read_csv('../data/vshape_30_clean_long.csv')
    print('Dataset loaded successfully.')
except FileNotFoundError:
    print('Error: Ensure data is in ../data/vshape_30_clean_long.csv')
```

## 1. Descriptive Statistics
Calculate the mean MTLD across the three authorial stages: Original Draft (OD), AI-Revised (AIR), and Author-Refined (AR).


```python
summary = df.groupby('Stage')['MTLD'].mean()
print('Mean MTLD by Stage:')
print(summary)

# Calculate percentage changes
od_val = summary['Original']
air_val = summary['AI_Revised']
ar_val = summary['Author_Refined']

print(f'\nReduction (OD to AIR): {((air_val - od_val)/od_val)*100:.2f}%')
print(f'Reclamation (AIR to AR): {((ar_val - air_val)/air_val)*100:.2f}%')
```

## 2. V-Shaped Trajectory Visualization
Plotting the trajectory across the three stages.


```python
plt.figure(figsize=(10, 6))
sns.pointplot(data=df, x='Stage', y='MTLD', order=['Original', 'AI_Revised', 'Author_Refined'], markers='o', linestyles='-')
plt.title('V-Shaped Trajectory in Lexical Diversity (MTLD)')
plt.ylabel('MTLD Score')
plt.xlabel('Authorial Stage')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```
